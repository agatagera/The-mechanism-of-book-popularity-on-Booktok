# Popularity and Market Success Indicators to create Engagement Metrics
* **view_count** - basic reach (number of times the video was viewed).
* **like_count** - initial engagement and acceptance of the content.
* **share_count** - key virality indicator (how often users shared the video).
* **favorites_count** - intention to return to the content/save recommendations for later.
* **comment_count** - discussion, building a community around the book.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_json('tiktok_booktok_2025_data.jsonl', lines=True)
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
# list of columns which will be use to calculate percentile ranks
metrics_columns = ['view_count', 'like_count', 'share_count', 'favorites_count', 'comment_count']
```

```python
# calculating percentile ranks for each metric
for col in metrics_columns:
    df[f'{col}_pct_rank'] = df[col].rank(pct=True, ascending=True)
df['raw_engagement_metrics'] = df[[f'{col}_pct_rank' for col in metrics_columns]].sum(axis=1)
```

```python
df['raw_engagement_metrics'].head()
```

```python
# check
print(df.sort_values(by='raw_engagement_metrics', ascending=False))
```

```python
max_possible_score = len(metrics_columns) * 1.0
df['engagement_metrics'] = (df['raw_engagement_metrics']/max_possible_score)* 100
```

```python
#check
df[['raw_engagement_metrics', 'engagement_metrics']].head()
```

```python
# check
print(df.sort_values(by='raw_engagement_metrics', ascending=False))
```
